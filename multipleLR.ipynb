```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
data_set = pd.read_csv(r'C:\Users\dell\Downloads\house_price_regression_dataset.csv')
X = np.column_stack((
    data_set['Square_Footage'],
    data_set['Num_Bedrooms'],
    data_set['Num_Bathrooms'],
    data_set['Year_Built'],
    data_set['Lot_Size'],
    data_set['Garage_Size'],
    data_set['Neighbourhood_Quality']
))
Y = data_set['House_Price'].values.reshape(-1, 1)
ones = np.ones((X.shape[0], 1), dtype=np.int32)
X = np.hstack((ones, X))
```

```python
#Now we calculate our parameter(beta) vector

beta = np.linalg.inv(np.transpose(X) @ X) @ (X.T @ Y)

y_bar = X@beta
print(y_bar.shape)
```

Testing our model

```python
import time

sqft = int(input("Enter number of square foot of the house"))
nbed = int(input('\nEnter number of bedrooms of the house'))
nbath = int(input('\nEnter number of Bathrooms'))
year_built = int(input("\nenter the year it was built "))
lot = float(input('\nEnter the lot size (in acres)'))
garage = int(input("\nEnter the Garage Size"))
neighborhood = int(input('\nEnter neighbourhood Quality (0 to 10)'))

print('\nPredicting the price')
time.sleep(1)
print('...')
time.sleep(1)
print('...')
time.sleep(1)
print('...')
time.sleep(1)
print('\nEstimating price based on previous data')
time.sleep(3)
test_input = np.array([1, sqft, nbed, nbath, year_built, lot, garage, neighborhood])
prediction = test_input @ beta

print(f"The predicted price of the house is {prediction}")
mean = np.mean(Y)
ssr_mean = np.sum((Y - mean)**2)
ssr_pred = np.sum((Y - y_bar)**2)
r_squared = (ssr_mean - ssr_pred)/ssr_mean
print(f"The model explains {r_squared * 100}% of variance in data")
```

Validation using sklearn

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
X = np.column_stack((
    data_set['Square_Footage'],
    data_set['Num_Bedrooms'],
    data_set['Num_Bathrooms'],
    data_set['Year_Built'],
    data_set['Lot_Size'],
    data_set['Garage_Size'],
    data_set['Neighbourhood_Quality']
))

Y = data_set['House_Price'].values.reshape(-1, 1)

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

X_train = np.hstack((np.ones((X_train.shape[0], 1)), X_train))
X_test = np.hstack((np.ones((X_test.shape[0], 1)), X_test))

beta = np.linalg.pinv(X_train) @ Y_train

Y_pred = X_test @ beta

r2 = r2_score(Y_test, Y_pred)
sqft = int(input("Enter number of square foot of the house"))
nbed = int(input('\nEnter number of bedrooms of the house'))
nbath = int(input('\nEnter number of Bathrooms'))
year_built = int(input("\nenter the year it was built "))
lot = float(input('\nEnter the lot size (in acres)'))
garage = int(input("\nEnter the Garage Size"))
neighborhood = int(input('\nEnter neighbourhood Quality (0 to 10)'))
print('\nPredicting the price')
time.sleep(1)
print('...')
time.sleep(1)
print('...')
time.sleep(1)
print('...')
time.sleep(1)
print('\nEstimating price based on previous data')
time.sleep(3)
test_input = np.array([1, sqft, nbed, nbath, year_built, lot, garage, neighborhood])
prediction = test_input @ beta

print(f"The predicted price of the house is {prediction}")

print(f"R² Score: {r2:.4f}")
```
